```python
# Import the tools we need.
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# Load a classic dataset: measurements of 150 flowers in 3 categories.
# It does not matter that they are flowers. What matters is the pattern:
# measurements go IN (the features), a category comes OUT (the label).
data = load_iris()
X = data.data       # the features: 4 measurements per example
y = data.target     # the label: which of 3 categories

print('Number of examples:', len(X))
print('Features per example:', X.shape[1])
print('Category names:', list(data.target_names))
```

```python
# Split the data: 70% to train on, 30% held back as an unseen test set.
# TODO: try changing test_size to 0.2 or 0.4 and re-run this cell.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42)

print('Training examples:', len(X_train))
print('Test examples (hidden from training):', len(X_test))
```

```python
# Train a k-Nearest Neighbors classifier.
# TODO: change n_neighbors (try 1, 3, 7, 15) and re-run this cell.
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)              # learn the boundary from the training set
knn_predictions = knn.predict(X_test)  # predict on the unseen test set
knn_accuracy = accuracy_score(y_test, knn_predictions)

print('k-NN accuracy on unseen data:', round(knn_accuracy * 100, 1), '%')
```

```python
# Train a Decision Tree classifier on the same split.
tree = DecisionTreeClassifier(random_state=42)
tree.fit(X_train, y_train)
tree_predictions = tree.predict(X_test)
tree_accuracy = accuracy_score(y_test, tree_predictions)

print('Decision Tree accuracy on unseen data:', round(tree_accuracy * 100, 1), '%')
```

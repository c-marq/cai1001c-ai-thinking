## What you are doing

In Chapter 6 you built a **generative** chatbot as a Gem, with no code. This lab builds the **other** kind — a **rule-based chatbot** — in Python.

A rule-based bot scans your message for **keywords** and returns a **scripted reply**. It has no understanding at all. Building one is the clearest way to see that — and to see exactly what a generative model adds, and risks.

Run each cell in order with **Shift + Enter**. This is pure Python — nothing to install.

```python
# A rule-based chatbot is a list of rules: (keywords, reply).
# The bot scans your message for any keyword and returns that rule's reply.
# TODO: add at least one new rule of your own to this list, then re-run this cell.
rules = [
    (["hour", "open", "close"], "The Student Help Desk is open Monday to Friday, 9 a.m. to 5 p.m."),
    (["password", "log in", "login"], "To reset your password, go to the campus IT portal and click 'Forgot password'."),
    (["advisor", "advising", "advise"], "Book an advising appointment in the student portal, under 'Advising'."),
    (["library", "book"], "The library is on the second floor and is open until 9 p.m. on weeknights."),
    (["bye", "goodbye", "thanks", "thank you"], "You're welcome. Good luck!"),
]

fallback = "Sorry, I don't have an answer for that. Try asking about hours, passwords, advising, or the library."

def respond(message):
    text = message.lower()                       # ignore capitalization
    for keywords, reply in rules:
        if any(word in text for word in keywords):
            return reply
    return fallback                              # no keyword matched

print('Chatbot ready. It knows', len(rules), 'rules.')
```

```python
# Test the bot with a few messages.
tests = [
    "What are your hours?",
    "I forgot my password",
    "How do I see an advisor?",
    "Do you sell coffee?",
]
for t in tests:
    print('You: ', t)
    print('Bot: ', respond(t))
    print()
```

```python
# Chat with your bot live. Type 'quit' to stop.
print("Help Desk Bot - type 'quit' to stop.")
while True:
    message = input('You: ')
    if message.lower().strip() == 'quit':
        print('Bot:  Goodbye!')
        break
    print('Bot: ', respond(message))
```
